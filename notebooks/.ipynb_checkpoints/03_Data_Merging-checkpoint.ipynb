```python
import pandas as pd
import numpy as np

DATA_PATH = "C:/Users/kalam/F1 Project Outcome Prediction/data/raw/"
PROCESSED_PATH = "C:/Users/kalam/F1 Project Outcome Prediction/data/processed/"
```

```python
races = pd.read_csv(DATA_PATH + "races.csv")
drivers = pd.read_csv(DATA_PATH + "drivers.csv")
results = pd.read_csv(DATA_PATH + "results.csv")
constructors = pd.read_csv(DATA_PATH + "constructors.csv")
status = pd.read_csv(DATA_PATH + "status.csv")
qualifying = pd.read_csv(DATA_PATH + "qualifying.csv")
pit_stops = pd.read_csv(DATA_PATH + "pit_stops.csv")
lap_times = pd.read_csv(DATA_PATH + "lap_times.csv")
```

```python
races = pd.read_csv(DATA_PATH + "races.csv")
drivers = pd.read_csv(DATA_PATH + "drivers.csv")
results = pd.read_csv(DATA_PATH + "results.csv")
constructors = pd.read_csv(DATA_PATH + "constructors.csv")
status = pd.read_csv(DATA_PATH + "status.csv")
qualifying = pd.read_csv(DATA_PATH + "qualifying.csv")
pit_stops = pd.read_csv(DATA_PATH + "pit_stops.csv")
lap_times = pd.read_csv(DATA_PATH + "lap_times.csv")
```

```python
races["date"] = pd.to_datetime(races["date"], errors="coerce")
drivers["dob"] = pd.to_datetime(drivers["dob"], errors="coerce")
drivers["driver_name"] = drivers["forename"].fillna("") + " " + drivers["surname"].fillna("")
drivers["driver_name"] = drivers["driver_name"].str.strip()
```

```python
pit_summary = pit_stops.groupby(["raceId", "driverId"], as_index=False).agg(
    pit_stop_count=("stop", "count"),
    mean_pit_stop_duration_ms=("milliseconds", "mean")
)

pit_summary.head()
```

```python
lap_summary = lap_times.groupby(["raceId", "driverId"], as_index=False).agg(
    avg_lap_time_ms=("milliseconds", "mean"),
    total_laps_recorded=("lap", "max")
)

lap_summary.head()
```

```python
quali = qualifying.copy()
quali = quali.rename(columns={"position": "quali_position"})
```

```python
f1 = results.merge(
    races[["raceId", "year", "round", "circuitId", "name", "date"]],
    on="raceId",
    how="left"
)

f1 = f1.merge(
    drivers[["driverId", "driver_name", "nationality", "dob"]],
    on="driverId",
    how="left"
)

f1 = f1.merge(
    constructors[["constructorId", "name", "nationality"]],
    on="constructorId",
    how="left",
    suffixes=("", "_constructor")
)

f1 = f1.merge(
    status,
    on="statusId",
    how="left"
)

f1 = f1.merge(
    quali[["raceId", "driverId", "constructorId", "quali_position", "q1", "q2", "q3"]],
    on=["raceId", "driverId", "constructorId"],
    how="left"
)

f1 = f1.merge(
    pit_summary,
    on=["raceId", "driverId"],
    how="left"
)

f1 = f1.merge(
    lap_summary,
    on=["raceId", "driverId"],
    how="left"
)
```

```python
f1 = f1.rename(columns={
    "name": "race_name",
    "name_constructor": "constructor_name",
    "nationality_constructor": "constructor_nationality",
    "status": "race_status"
})
```

```python
selected_cols = [
    "resultId", "raceId", "year", "round", "race_name", "date", "circuitId",
    "driverId", "driver_name", "nationality", "dob",
    "constructorId", "constructor_name", "constructor_nationality",
    "grid", "position", "positionOrder", "points", "laps",
    "fastestLap", "fastestLapTime", "fastestLapSpeed",
    "quali_position", "q1", "q2", "q3",
    "pit_stop_count", "mean_pit_stop_duration_ms",
    "avg_lap_time_ms", "total_laps_recorded",
    "race_status"
]

f1_merged = f1[selected_cols].copy()
f1_merged.head()
```

```python
f1_merged.to_csv(PROCESSED_PATH + "f1_merged.csv", index=False)
print("Saved: f1_merged.csv")
print("Shape:", f1_merged.shape)
```
