```python
import pandas as pd
import matplotlib.pyplot as plt

PROCESSED_PATH = "C:/Users/kalam/F1 Project Outcome Prediction/data/processed/"
f1 = pd.read_csv(PROCESSED_PATH + "f1_merged.csv")
```

```python
f1["date"] = pd.to_datetime(f1["date"], errors="coerce")
f1["position"] = pd.to_numeric(f1["position"], errors="coerce")
f1["positionOrder"] = pd.to_numeric(f1["positionOrder"], errors="coerce")
f1["grid"] = pd.to_numeric(f1["grid"], errors="coerce")
```

```python
## Top drivers by Wins
```

```python
wins = f1[f1["positionOrder"] == 1]
top_drivers = wins["driver_name"].value_counts().head(10)

plt.figure(figsize=(10, 6))
top_drivers.plot(kind="bar")
plt.title("Top 10 Drivers by Race Wins")
plt.xlabel("Driver")
plt.ylabel("Wins")
plt.xticks(rotation=45)
plt.show()
```

```python
## Top Constructors by Wins
```

```python
top_teams = wins["constructor_name"].value_counts().head(10)

plt.figure(figsize=(10, 6))
top_teams.plot(kind="bar")
plt.title("Top 10 Constructors by Race Wins")
plt.xlabel("Constructor")
plt.ylabel("Wins")
plt.xticks(rotation=45)
plt.show()
```

```python
## Number of Races per Season
```

```python
races_per_year = f1[["year", "raceId"]].drop_duplicates()["year"].value_counts().sort_index()

plt.figure(figsize=(12, 6))
races_per_year.plot(kind="line", marker="o")
plt.title("Number of Races per Season")
plt.xlabel("Year")
plt.ylabel("Race Count")
plt.show()
```

```python
## Grid Position vs Finishing Position
```

```python
plot_df = f1[["grid", "positionOrder"]].dropna()

plt.figure(figsize=(8, 6))
plt.scatter(plot_df["grid"], plot_df["positionOrder"], alpha=0.3)
plt.title("Grid Position vs Finishing Position")
plt.xlabel("Grid Position")
plt.ylabel("Finishing Position")
plt.show()
```

```python
## Top drivers by Podium
```

```python
podiums = f1[f1["positionOrder"] <= 3]
top_podiums = podiums["driver_name"].value_counts().head(10)

plt.figure(figsize=(10, 6))
top_podiums.plot(kind="bar")
plt.title("Top 10 Drivers by Podium Finishes")
plt.xlabel("Driver")
plt.ylabel("Podiums")
plt.xticks(rotation=45)
plt.show()
```
