```python
# 2. Data Cleaning

This notebook checks:
- missing values
- duplicates
- incorrect data types
- basic cleaning steps required before merging
```

```python
import pandas as pd
import numpy as np

DATA_PATH = "C:/Users/kalam/F1 Project Outcome Prediction/data/raw/"

races = pd.read_csv(DATA_PATH + "races.csv")
drivers = pd.read_csv(DATA_PATH + "drivers.csv")
results = pd.read_csv(DATA_PATH + "results.csv")
qualifying = pd.read_csv(DATA_PATH + "qualifying.csv")
constructors = pd.read_csv(DATA_PATH + "constructors.csv")
status = pd.read_csv(DATA_PATH + "status.csv")
```

```python
tables = {
    "races": races,
    "drivers": drivers,
    "results": results,
    "qualifying": qualifying,
    "constructors": constructors,
    "status": status
}
```

```python
## Missing values
```

```python
for name, df in tables.items():
    print(f"\n{name.upper()} - Missing Values")
    print(df.isnull().sum())
```

```python
## Duplicate Rows
```

```python
for name, df in tables.items():
    print(f"{name} duplicate rows:", df.duplicated().sum())
```

```python
## Data Type Fixes
```

```python
races["date"] = pd.to_datetime(races["date"], errors="coerce")
drivers["dob"] = pd.to_datetime(drivers["dob"], errors="coerce")
```

```python
drivers["driver_name"] = drivers["forename"].fillna("") + " " + drivers["surname"].fillna("")
drivers["driver_name"] = drivers["driver_name"].str.strip()

drivers[["driverId", "driver_name", "nationality"]].head()
```
